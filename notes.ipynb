# File Processing

#### File Paths:
-   Absolute Paths - Beings at the root folder. Starts at the base of your computer.
    -   ex: `/Volumes/ExtHD/PerScholas/RTT-36/385/385.3/notes/notes.ipynb`
-   Relative paths - Path is relative to the current (working) directory
    -   ex: `./notes.ipynb`

#### File Path Symbols
-   `/` - divider between folders/files
-   `.` - currect directory
-   `..` - One directory up the chain

#### Opening Files
-   `open(file_path, mode)` - function

#### File modes
-   `r` - read - if file does not exists (FileNotFoundError)
-   `w` - write - if file doesnt exist, it creates a new file
-   `a` - append - if file doesnt exist, it creates a new file
-   `x` - exlusive creation - if file doesnt exist, it creates a new file
-   `b` - binary
-   `t` - text
-   `+` - read/write

```python
f = open('data/my_file.txt', mode='r') # opens the file

print(f.read()) # reads the file

f.close() # Always close your file to prevent extra resources and breaking of files.
```

#### Error Handling w/ Try: Except: Finally
-   `try:` - this block is code we want to run
-   `except:` - raise exceptions for errors we run into. Code to run if we get an error
-   `finally:` - code we run no matter what happens. errors or not

```python
try:
    f = open('data/my_file.txt', mode='r')

    print(f.read())

except Exception as e: # if an error, put in variable 'e' and print error to logs (terminal)
    print(f'❌ Error: {e}')

finally:
    f.close()
```

```python
# if opening file with 'with', all code goes into the with code block and the file auto closes at the end of the block
with open('data/my_file.txt', 'r') as f:
    print(f.read())
```

```python
def write_to_file(str, file_path):
    try:
        with open(file_path, 'w') as f:
            f.write(str + '\n')
            
        print('Success')
    except Exception as e:
        print(f'❌ Error: {e}')
```

```python
def append_to_file(str, file_path):
    try:
        with open(file_path, 'a') as f:
            f.write(str + '\n')
            
        print('Success')
    except Exception as e:
        print(f'❌ Error: {e}')
```

```python
# Define a list of dishes on the menu
menu = [
    "Spaghetti Carbonara",
    "Chicken Alfredo",
    "Margherita Pizza",
    "Cheeseburger",
    "Caesar Salad",
    "Fish and Chips",
    "Pad Thai",
    "Sushi Platter",
    "Vegetable Stir-Fry",
    "Grilled Salmon"
]

# Function to suggest a random dish
def suggest_dish():
    random_dish = random.choice(menu)
    return random_dish

# Main program
print("Welcome to the Restaurant Menu Randomizer!")
print("If you're undecided about what to order, let us help you decide.")

while True:
    user_input = input("Press enter to get a random dish suggestion (or type 'quit' to exit): ")

    if user_input.lower() == 'quit':
        print("Thank you for using the Restaurant Menu Randomizer. Enjoy your meal!")
        break

    suggested_dish = suggest_dish()
    print("Randomly suggested dish:", suggested_dish)
```

```python
arr = np.array([1, 2, 3, 4, 5])

element = arr[3]
print('Element: ', element)

slice = arr[1:4]
print('Slice: ', slice)

print(arr)
```

```python
# Arithmetic and NumPy Arrays
# Arithmetic with arrays and scalar values compute element-wise (PER ELEMENT, for each element)
arr = np.array([1, 2, 3])
arr2 = np.array([9, 8, 7])

print(arr + 1) #[2, 3, 4]
print(arr * 3) #[3, 6, 9]
print(arr / 2) # [0.5, 1.0, 1.5]

print(arr + arr2) # [10, 10, 10] added elements at the same index together.
```

```python
# NumPy Aggregations
arr = np.array([1, 2, 3, 4, 5])

sum = np.sum(arr)
mean = np.mean(arr)
median = np.median(arr)
stDev = np.std(arr)
min = np.min(arr)
max = np.max(arr)

print('Sum: ', sum)
print('Avg: ', mean)
print('Median: ', median)
print('Standard Deviation: ', stDev)
print('Min: ', min)
print('Max: ', max)
```

```python
# Logical Operators AND, OR NOT
a = np.array([True, True, False, False])
b = np.array([True, False, True, False])
print('Logical And') # compare each element in a list
print(np.logical_and(a, b))

print('\nLogical Or') # compare each element in a list
print(np.logical_or(a, b))

c = np.array([True, True, False, False])
print(np.logical_not(c)) # Flips the boolean value

a = np.array([1, 2, 3, 4])
b = np.array([1, 2, 3, 4])
c = np.array([2, 3, 4, 5])

print(np.array_equal(a, b)) # compares the entire array looking for differences.

print(np.array_equal(a, c))
```
