```python
%pip install matplotlib
import numpy as np
import matplotlib.pyplot as plt

# Define base data points
xpoints = np.array([1, 8])
ypoints = np.array([3, 10])

# Create a figure to hold all examples
plt.figure(figsize=(8, 6))

# Example 1: Dashdot Line style ('dashdot' or '-.')
plt.plot(xpoints, ypoints, linestyle="dashdot", label="Dashdot", color="blue")

# Example 2: Dotted Line style (':' or 'dotted')
# Offsetting y by 2 to visualize on the same plot
plt.plot(xpoints, ypoints + 2, linestyle=":", label="Dotted", color="red")

# Example 3: Dash Line style ('--' or 'dashed')
# Offsetting y by 4
plt.plot(xpoints, ypoints + 4, linestyle="--", label="Dashed", color="green")

# Example 4: Solid Line style ('-' or 'solid')
# Offsetting y by 6
plt.plot(xpoints, ypoints + 6, linestyle="-", label="Solid", color="purple")

# Add labels and legend for clarity (good practice for grading)
plt.title("Lab 386.5.1 - Plotting Line Styles")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.legend()

# Display the plot
plt.show()
```

```python
x = np.arange(1, 12, 2)
y = np.arange(2, 13, 2)

# create a line plot with a dashdot linestyle
plt.plot(x, y, linestyle='-.', marker='*', markersize='20')
plt.show()
```

```python
fig, ax = plt.subplots()

ax.plot(x, y, linestyle='-.', marker='*')
plt.show()
```

```python
plt.plot(x, y, linestyle=':')
plt.show()
```
