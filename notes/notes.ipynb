# Datetime modules
-   `datetime` - built-in module for datetime functionality
-   `calendar` - built-in module, calendar related functionality
-   `time` - module for time functionality

```python
from datetime import date, datetime, time
```

```python
# Create date object
last_year = date(2025, 9, 18)
print(last_year)
```

```python
today_to_the_sec = datetime.now()
print(today_to_the_sec)
print(today_to_the_sec.year)
print(today_to_the_sec.month)
print(today_to_the_sec.day)
print(today_to_the_sec.hour)
print(today_to_the_sec.minute)
print(today_to_the_sec.second)
print(today_to_the_sec.microsecond)
print(today_to_the_sec.tzinfo)

print(today_to_the_sec.timestamp())
```

```python
time_ish = time(hour=12, minute=45, second=45)
print(time_ish)
```

```python
import numpy as np

print('Numpy Version: ', np.__version__)
```

```python
one_to_100 = np.arange(100)

print(one_to_100)

# array from 1 - 30 not including 30
one_thirty = np.arange(1, 30)
print(one_thirty)

# array from 10-20 by 2
ten_twenty = np.arange(10, 20, 2)
print(ten_twenty)

# array from -3 -10 10 by 4, verify dtype is int
print(np.arange(-3, 10, 4, dtype=int))
```

```python
from numpy import random

# Generate a random array of 10 numbers between 0-1
data = random.rand(10)

print(data)

# Generate an array of random numbers with a normal distribution
normal_dist = random.randn(10)

print(normal_dist)

# Normally distributed Matrix, 3 rows, 2 records in each row
normal_dist_matrix = random.randn(3, 2)

print(normal_dist_matrix)
```

```python
# Define a list
elements = ['a', 'b', 'c', 'd', 'e', 'f']

# Choose a random value from the list
random_sample = random.choice(elements)

print(random_sample)

print('Before Shuffle', elements)
random.shuffle(elements)
print('After Shuffle', elements) # shuffling elements of array for randomness
```
