```python
import pandas as pd

data = {
    'Category': ['A', 'A', 'B', 'B', 'A', 'B','A','B'],
    'Value': [10, 15, 20, 25, 30, 35,40,50],
    'Fee' :[20000,25000,26000,22000,24000,35000,4000,5600],
    'Duration':['30day','40days','35days','40days','60days','60days','50days','60days'],
    'Discount':[1000,2300,1200,2500,2000,2000,500,600]
}
df = pd.DataFrame(data)
print("Base DataFrame:")
print(df)
```

```python
# Calculate Total Sum
print("Total Sum:\n", df[['Value']].sum())

# Calculate Cumulative Sum
print("\nCumulative Sum:\n", df[['Value']].cumsum())

# Calculate Mean (Average)
print("\nAverage Value:\n", df[['Value']].mean())

# Calculate Maximum Value
print("\nMaximum Value:\n", df[['Value']].max())

# Calculate Total Count
print("\nTotal Count:\n", df[['Value']].count())
```

```python
# Multiple aggregations on specific columns
print("Total Fee and Discount:\n", df[['Fee','Discount']].sum())

# Grouping by Category to see sums per group
result_Group = df.groupby('Category')[['Value', 'Fee', 'Discount']].aggregate('sum')
print("\nGrouped by Category (Sums):\n", result_Group)

# Multi-Column Summary Statistics (Min, Max, Mean)
result_Group2 = df[['Fee', 'Discount']].aggregate(['min', 'max', 'mean'])
print("\nSummary Statistics:\n", result_Group2)
```
